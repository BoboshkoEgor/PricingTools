# 02 — Campaign Funnel & Channel Metrics

Goal: compare campaign funnel metrics across simulated email, SMS and push campaigns. Use clicks and conversions as decision-grade outcomes; email opens are included as a diagnostic only.

## 1. Simulate campaign counts

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
np.random.seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
campaigns=pd.DataFrame([
 {'channel':'Email','sent':80000,'delivered':77600,'opens':25000,'clicks':3100,'conversions':620,'unsub':155},
 {'channel':'SMS','sent':18000,'delivered':17460,'opens':np.nan,'clicks':1450,'conversions':510,'unsub':82},
 {'channel':'Push','sent':60000,'delivered':54000,'opens':np.nan,'clicks':2700,'conversions':430,'unsub':240},
])
campaigns
```

## 2. Calculate funnel rates
Definitions follow the lecture: delivery / sent; email open / delivered; CTR = clicks / delivered; CTOR = clicks / opens; conversion = conversions / delivered; unsubscribe = unsubscribes / delivered.

```python
campaigns['delivery_rate']=campaigns.delivered/campaigns.sent
campaigns['open_rate']=campaigns.opens/campaigns.delivered
campaigns['CTR']=campaigns.clicks/campaigns.delivered
campaigns['CTOR']=campaigns.clicks/campaigns.opens
campaigns['conversion_rate']=campaigns.conversions/campaigns.delivered
campaigns['unsubscribe_rate']=campaigns.unsub/campaigns.delivered
rate_cols=['delivery_rate','open_rate','CTR','CTOR','conversion_rate','unsubscribe_rate']
campaigns[rate_cols].applymap(lambda x: f'{x:.2%}' if pd.notna(x) else '—')
```

## 3. Funnel visualization (email)

```python
email=campaigns.loc[campaigns.channel.eq('Email')].iloc[0]
funnel=pd.Series({'Sent':email.sent,'Delivered':email.delivered,'Opened':email.opens,'Clicked':email.clicks,'Converted':email.conversions})
funnel.plot(kind='bar',figsize=(8,4),title='Email campaign funnel',ylabel='Unique customers/events',color='#8b5cf6')
plt.xticks(rotation=20); plt.tight_layout(); plt.show()
```

## 4. Channel comparison
SMS has no open metric in this simulation. Do not compare open rates across channels or interpret email opens as fully reliable (privacy prefetch can inflate them).

```python
campaigns.set_index('channel')[['CTR','conversion_rate','unsubscribe_rate']].plot(kind='bar',figsize=(9,4))
plt.title('Campaign rates by channel'); plt.ylabel('Rate'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

## 5. Short report
Read the metrics together: channel cost, audience quality, conversion and unsubscribe behavior matter. CTR is clicks / delivered; CTOR is clicks / opens. A campaign with a high conversion rate among recipients is not necessarily incremental lift; see notebook 03.
